# Lindholmen Dataset

The Lindholmen dataset is used to source activity diagrams to redraw as sketch.

## Libraries

```python
import pandas as pd
from numpy import nan
from urllib.parse import urlparse
import requests
from tqdm import tqdm
import time
```

## Methods

```python
def get_url_extension(url):
    if pd.isna(url):
        return ''
    # Parse the URL and get the path
    path = urlparse(str(url)).path
    # Extract extension from path
    if '.' in path:
        return path.split('.')[-1].lower()
    return ''
```

## Code

Load CSV file

```python
dataset = pd.read_csv("UMLFiles_List_V2.0.csv")
```

```python
dataset.columns
```

```python
dataset.rename(columns={'Model Link - Github': 'Link'}, inplace=True)
```

Isolate username and project. Extract file extension

```python
dataset['Username'] = dataset['Project'].str.split("/").str[0]
dataset['Repository'] = dataset['Project'].str.split("/").str[1]
dataset['File_Extension'] = dataset['Link'].apply(get_url_extension)
```

Reorder columns

```python
dataset = dataset[['Username', 'Repository', 'Project', 'Link', 'File_Extension']]
```

Inspect file extensions

Remove all files that does not have the extension "png", "jpg", "svg", "jpeg", "gif" or "bmp"

```python
dataset = dataset[dataset['File_Extension'].isin(['png', 'jpg', 'svg', 'jpeg', 'gif', 'bmp'])].reset_index(drop=True)
```
